```python
#Space Resource Calculations
travel_time = 10  # hours
speed = 5000      # kilometers per hour
fuel_efficiency = 10  # kilometers per liter
first_name = "Sally"
last_name = "Ride"
Distance = speed*travel_time
Fuel_Consumption = Distance/fuel_efficiency
Time_to_Destination = Distance/speed
full_name = first_name + " " + last_name
print("Total Distance", Distance)
print("Fuel Consumed", Fuel_Consumption)
print("Total Time", Time_to_Destination)
print("Full Name", full_name)
```

```python
#Satellite Mass Calculator
structure = 950 #kg
propulsion_system = 420 #kg
communication_payload = 135 #kg
power_subsystem = 90 #kg
total_mass = structure + propulsion_system + communication_payload + power_subsystem
contingency_reserve = total_mass * 5/100
final_mass = total_mass + contingency_reserve
print("Total","", "mass", total_mass)
print("Contingency","", "reserve", contingency_reserve)
print("Final mass including contingency reserve", final_mass)
```
